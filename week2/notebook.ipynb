```python
#AUTOR: MIGUEL ANGEL CORDOBA FIGUEROA

# Esta función recibe:
# x -> lista de entradas
# w -> lista de pesos
# b -> sesgo (bias), que ajusta el umbral de activación
def perceptron(x, w, b):
    # Cálculo del puntaje z: suma de (entrada * peso) + bias
    z = sum(x_i * w_i for x_i, w_i in zip(x, w)) + b

    # Función de activación: salida binaria
    y = 1 if z >= 0 else 0
    return z, y

# Entradas controladas: todas las combinaciones de dos bits
entradas = [[0,0],[0,1],[1,0],[1,1]]

# Definimos los 10 casos de prueba con pesos y bias                      ¿QUE SE BUSCA DEMOSTRAR EN CADA CASO?
casos_prueba = [
    ("Prueba 1: variación de pesos", [1.0, 0.5], -0.7),                          # Mostrar cómo un peso mayor en una entrada domina la decisión del perceptrón.
    ("Prueba 2: variación del bias", [0.5, 0.5], 0.5),                           # Evidenciar que un bias positivo facilita la activación incluso con entradas bajas.
    ("Prueba 3: compuerta OR", [0.5, 0.5], -0.2),                                # Demostrar que el perceptrón puede simular una compuerta OR.
    ("Prueba 4: bias más negativo", [0.5, 0.5], -1.0),                           # Ver cómo un bias negativo fuerte hace más estricta la activación.
    ("Prueba 5: pesos distintos, bias neutro", [0.8, 0.2], 0.0),                 # Resaltar que un peso dominante puede controlar la salida aunque el bias sea neutro.
    ("Prueba 6: pesos altos, bias fuerte negativo", [1.0, 1.0], -1.5),           # Comprobar que un bias negativo puede contrarrestar pesos grandes.
    ("Prueba 7: compuerta AND clásica", [0.5, 0.5], -0.7),                       # Reproducir el comportamiento clásico de una compuerta AND.
    ("Prueba 8: compuerta NAND", [-0.5, -0.5], 0.7),                             # Mostrar que con pesos negativos y bias positivo se obtiene una compuerta NAND.
    ("Prueba 9: compuerta NOR", [-0.5, -0.5], 0.2),                              # Evidenciar que el perceptrón puede simular una compuerta NOR.
    ("Prueba 10: compuerta NOT (una entrada)", [-1.0], 0.5)                      # Demostrar que con una sola entrada y peso negativo se logra una compuerta NOT.
]

# Ejecución de todas las pruebas
for nombre, pesos, bias in casos_prueba:
    print("\n" + nombre)
    print("x1 x2 | z | salida")
    for x in entradas:
        # Ajuste para pruebas con una sola entrada (NOT)
        if len(pesos) == 1:
            z, y = perceptron([x[0]], pesos, bias)
            print(f"{x[0]}     | {z:.2f} | {y}")
        else:
            z, y = perceptron(x, pesos, bias)
            print(f"{x[0]}  {x[1]}  | {z:.2f} | {y}")
```

  **Análisis y conclusiones sobre los parámetros**

 **Prueba 1**: Un peso mayor en una entrada domina la decisión del perceptrón.  
 **Prueba 2**: Un bias positivo facilita la activación incluso con entradas bajas.  
 **Prueba 3**: El perceptrón puede simular una compuerta OR.  
 **Prueba 4**: Un bias negativo fuerte hace más estricta la activación.  
 **Prueba 5**: Un peso dominante puede controlar la salida aunque el bias sea neutro.  
 **Prueba 6**: Un bias negativo puede contrarrestar incluso pesos grandes.  
 **Prueba 7**: El perceptrón reproduce el comportamiento clásico de una compuerta AND.  
 **Prueba 8**: Con pesos negativos y bias positivo se obtiene una compuerta NAND.  
 **Prueba 9**: El perceptrón simula una compuerta NOR.  
 **Prueba 10**: Con una sola entrada y peso negativo se logra una compuerta NOT.  



### Conclusión general
Los pesos determinan la importancia de cada entrada en la decisión, mientras que el bias ajusta el umbral de activación. Al variar estos parámetros, el perceptrón puede comportarse como diferentes compuertas lógicas (AND, OR, NAND, NOR, NOT), evidenciando su versatilidad y la base de cómo funcionan redes neuronales más complejas.
